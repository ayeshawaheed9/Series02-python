## Plot a sinusoidal wave Final version

```python
import numpy as np
import matplotlib.pyplot as plt

# --------------------------------------------------
# PARAMETERS
# --------------------------------------------------

omega0 = 1
alpha = 0.3

# --------------------------------------------------
# PHASE PLANE
# x = displacement
# v = velocity
# --------------------------------------------------

x = np.linspace(-4, 4, 20)
v = np.linspace(-4, 4, 20)

X, V = np.meshgrid(x, v)

# System:
# x' = v
# v' = -omega0^2*x - 2*alpha*omega0*v

X_dot = V
V_dot = -omega0**2 * X - 2*alpha*omega0 * V

# Normalize arrows so they have similar length
speed = np.sqrt(X_dot**2 + V_dot**2)
U = X_dot / speed
W = V_dot / speed

# --------------------------------------------------
# DRAW
# --------------------------------------------------

fig, ax = plt.subplots(figsize=(8, 8))

# Direction field
ax.quiver(
    X, V, U, W,
    angles="xy",
    pivot="mid",
    alpha=0.45
)

# --------------------------------------------------
# NULLCLINES
# --------------------------------------------------

# x-nullcline:
# x' = 0  ->  v = 0
ax.axhline(
    0,
    linewidth=2,
    label="x-nullcline: x' = 0  →  v = 0"
)

# v-nullcline:
# v' = 0  ->  v = -(omega0/(2*alpha))*x
x_line = np.linspace(-4, 4, 200)
v_line = -(omega0 / (2 * alpha)) * x_line

ax.plot(
    x_line,
    v_line,
    linewidth=2,
    label="v-nullcline: v' = 0"
)

# --------------------------------------------------
# EQUILIBRIUM
# --------------------------------------------------

ax.plot(
    0, 0,
    marker="o",
    markersize=9,
    label="Equilibrium (0,0)"
)

# --------------------------------------------------
# TRAJECTORIES
# --------------------------------------------------

# Starting points
initial_conditions = [
    (3, 0),
    (2, 2),
    (-3, 0),
    (-2, -2),
    (1, -3),
    (-1, 3)
]

# Simple numerical simulation
dt = 0.02
steps = 1500

for x0, v0 in initial_conditions:

    xs = [x0]
    vs = [v0]

    xx = x0
    vv = v0

    for _ in range(steps):

        dx = vv
        dv = -omega0**2 * xx - 2*alpha*omega0*vv

        xx = xx + dx * dt
        vv = vv + dv * dt

        xs.append(xx)
        vs.append(vv)

    ax.plot(xs, vs, linewidth=1.5)

# --------------------------------------------------
# LABELS
# --------------------------------------------------

ax.set_xlabel("x (displacement)", fontsize=13)
ax.set_ylabel("v (velocity)", fontsize=13)

ax.set_title(
    "Phase Portrait of the Damped Landing Model",
    fontsize=14
)

ax.set_xlim(-4, 4)
ax.set_ylim(-4, 4)

ax.axvline(0, linewidth=0.8, alpha=0.3)
ax.axhline(0, linewidth=0.8, alpha=0.3)

ax.grid(True, alpha=0.2)
ax.legend()

plt.show()
```

Formula to generate a sine wave is x(t) = A sin(ωt +  φ)

- A is the amplitude
- t is the time
- ω is the angular frequency which is equal to 2 * pi * f (where f is the frequency)
- φ is the angle in radian

```python
# Create a new code cell in the notebook, in which you will:
# Import the needed libraries
import numpy as np
import matplotlib.pyplot as plt
# Create a numpy array called t that contains 1000 values from 0 to 1 (t is for time)
t = np.linspace(0, 1, 1000)
# Create a numpy array called s that contains the values of a sine function of t (s is for signal)
#Amplitude of the signal
A=1
#Frequency of the signal
f=1
# ω = 2 * np.pi * f
omega = 2 * np.pi * f

# Create a function called sinusoid that : 
def sinusoid(A, f, t, phase):
    """
    Compute the values of a sinusoidal signal x(t) = A * sin(ω t + φ).

    Parameters
    ----------
    A : 
        Peak amplitude (A) of the signal.
    f :
        Frequency in Hz.
    t : 
        Time values (in seconds) at which to evaluate the signal.
    phase : 
        Phase offset in radians.

    Returns
    -------
    numpy.ndarray
        The signal values evaluated at each point in `time`.
    """
    omega = 2 * np.pi * f
    x = A * np.sin(omega * t + phase)
    return x

# - Use the function sinusoid to create two numpy arrays s1 and s2 with the two different frequencies
s1 = sinusoid(A, 1, t, 0)      # f = 1 → one full cycle
s2 = sinusoid(A, 2, t, 0)    # f = 2 → faster oscillation 

# - Plot s1 and s2 as a function of t (as illustrated in the graph below).
plt.plot(t, s1, label="Sinus1 (1 Hz)")
plt.plot(t, s2, label="Sinus2 (2 Hz)")
# Add a title to the plot
plt.title("Sinusoidal Functions")
# Add a label to the x axis
plt.xlabel("Time")
# Add a label to the y axis
plt.ylabel("Amplitude")
# Add a grid to the plot
plt.grid(True)
# Add a legend to the plot
plt.legend()
# Save the plot as a pdf file
plt.savefig("sine_waves_frequencies.pdf", format="pdf")


# gave prompt to claude only to see how to create the numpy array which the sine function has to return.
# rest of it is done by me with little help from copilot.
```
