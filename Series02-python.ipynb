## Plot a sinusoidal wave better version

Formula to generate a sine wave is x(t) = A sin(ωt +  φ)

- A is the amplitude
- t is the time
- ω is the angular frequency which is equal to 2 * pi * f (where f is the frequency)
- φ is the angle in radian

```python
# Create a new code cell in the notebook, in which you will:
# Import the needed libraries
import numpy as np
import matplotlib.pyplot as plt
# Create a numpy array called t that contains 1000 values from 0 to 1 (t is for time)
t = np.linspace(0, 1, 1000)
# Create a numpy array called s that contains the values of a sine function of t (s is for signal)
#Amplitude of the signal
A=1
#Frequency of the signal
f=1
# ω = 2 * np.pi * f
omega = 2 * np.pi * f
s = A * np.sin(omega * t + 2*np.pi)  # Phase shift of 2π
# Create a numpy array called c that contains the values of a cosine function of t (c is for cosine signal)
c = A * np.cos(omega * t + 2*np.pi)  # Phase shift of 2π
# Plot s as a function of t
plt.plot(t, s, label="Sine")
plt.plot(t, c, label="Cosine")
# Add a title to the plot
plt.title("Sine and Cosine Waves")
# Add a label to the x axis
plt.xlabel("Time")
# Add a label to the y axis
plt.ylabel("Amplitude")
# Add a grid to the plot
plt.grid(True)
# Add a legend to the plot
# what is legend? A legend is a small box that explains the meaning of the different colors or symbols used in a plot. It helps the reader understand what each line or point represents. In this case, the legend will show which line corresponds to the sine wave and which line corresponds to the cosine wave.
plt.legend()    
# Save the plot as a pdf file
plt.savefig("sine_cosine_wave.pdf", format="pdf")



# - Add a second plot of a cosine function of t in the same figure

#  - Add a legend to the plot
#  - Add a second y axis to the plot
#  - Add a second label to the y axis
#  - Save the plot as a pdf file
```
